- **Target/Proxy:** `high_visibility` (1 if organic traffic/visibility exceeds threshold, 0 otherwise).

```python
# 4. The Unit of Analysis as a Real DataFrame
import pandas as pd
import numpy as np

np.random.seed(42)
df = pd.DataFrame({
    'page_id': [f'page_{i}' for i in range(5)],
    'content_length': [1200, 450, 2100, 800, 1500],
    'keyword_density': [2.1, 0.8, 3.0, 1.2, 2.5],
    'high_visibility': [1, 0, 1, 0, 1]
})
print("Unit of Analysis: One row = One Web Page")
print(df)
```
