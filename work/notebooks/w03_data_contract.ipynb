## 4. The Leakage Trap & Limitation
- **Leakage Experiment:** Adding a future outcome label directly as a feature causes artificial performance jumps, which must be strictly removed.
- **Named Limitation:** Analysis is restricted to sample structural features and does not capture live search engine algorithm updates.
